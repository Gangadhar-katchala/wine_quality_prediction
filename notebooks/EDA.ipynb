```python
def quality_labeler(quality):
    if quality <= 4:
        return 'low'
    elif quality <= 6:
        return 'medium'
    else:
        return 'high'
```

```python
df_scaled['quality_label'] = df_scaled['quality'].apply(quality_labeler)
df_scaled.drop(columns=['quality'], inplace=True)
df_scaled.head()
```

```python
import matplotlib.pyplot as plt

scores_df = pd.DataFrame(list(scores.items()), columns=["Model", "Score"])
scores_df.sort_values(by="Score", ascending=False, inplace=True)

plt.figure(figsize=(10, 6))
bars = plt.barh(scores_df["Model"], scores_df["Score"], color='skyblue')
plt.xlabel("Score")
plt.title("Model Comparison (Test Score)")
for bar in bars:
    plt.text(
        bar.get_width() + 0.01,               
        bar.get_y() + bar.get_height() / 2,    
        f"{bar.get_width():.4f}",              
        va='center'
    )

plt.xlim(0, 1)  
plt.show()
```
